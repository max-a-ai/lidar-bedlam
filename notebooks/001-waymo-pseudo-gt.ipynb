# Are the Waymo pseudo-GT meshes good enough to train on?

Waymo labels pedestrians with 3D keypoints and boxes, not meshes. The
fitter in `lidar_bedlam/generate/pseudo_fit_v2.py` produces SMPL labels for
them; v3 (`real/v1_pseudo3`) adds a hip-centre term and per-keypoint
convention offsets. Cell bodies:
[ipynb-assets/nb001_waymo_pseudo_gt.py](ipynb-assets/nb001_waymo_pseudo_gt.py).

## Checks
- [x] v3 labels: 4,145 accepted fits, 31 mm keypoint error, hip shift 7 mm (was 42 / 35 mm)
- [x] Every label set (v3, v2, LiDAR-HMR, pedestrian generation, v1) compared on the same records
- [x] LiDAR-HMR's own published Waymo labels inspected on their returns and keypoints

**Verdict:** v3 is the best label set we have and trains straight wrists (9° bend, m-boxhead-v3); use `real/v1_pseudo3`.
