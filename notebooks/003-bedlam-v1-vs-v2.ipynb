BEDLAM 1.0 is the training source (clothed depth, masks, SMPL labels).
The 2.0 download on the NAS is compared on cameras, motion, bodies and

## Checks
- [x] Inventory of both downloads: frames, PNG, depth, masks, body ground truth
- [x] Camera height, pitch and FOV against the car-roof, head-mounted, pole and drone rigs
- [x] Motion configs, frames, overview sheets, bodies and assets per release
- [x] Both LiDAR routes on one 1.0 frame: clothed depth map vs SMPL mesh plus 2 cm

**Verdict:** 2.0 adds moving cameras, shake and 0.5–2.5 m heights; neither release covers pole or drone views; the 2.0 download has no depth, masks or body GT, so 1.0 stays the training source.

- 2.0 cameras sit 0.5–2.5 m high, moving and shaking: ego and head-mounted domains, not cars.
- 1.0 closeup and stadium groups already reach 5–6 m at 15–37° down: infrastructure at 5 m is covered.
- Neither release has frames above 5.6 m or steeper than 40°: no pole (7 m) or drone views.
- The 2.0 download holds PNG and camera CSVs only: depth, masks and body GT were not released here.
- Mixing 2.0 in needs its body GT and a mesh-based LiDAR route; worth it only for ego views.
